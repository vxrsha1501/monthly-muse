# 01 — Cosine calibration

Calibrates the similarity thresholds used by MonthlyMuse's NLP engine (Section 6.5):
embed pairs of messages with the own 384-d embedder, plot the cosine distribution for
near-duplicates vs. unrelated posts, and pick the repetition-alert threshold (cos >= 0.85).

```python
import sys, pathlib, numpy as np
sys.path.insert(0, str(pathlib.Path('..', 'backend').resolve()))

from app.ai.embedder import get_embedder
from app.ai.similarity import cosine

emb = get_embedder()
pairs = [
    ("Fresh pastries every morning at Brew & Bean!",
     "Every morning we bake fresh pastries at Brew and Bean."),   # near-duplicate
    ("Fresh pastries every morning at Brew & Bean!",
     "Our November newsletter: three tips for better sleep."),      # unrelated
]
for a, b in pairs:
    print(f"cosine = {cosine(emb.embed([a])[0], emb.embed([b])[0]):.3f}")
```
